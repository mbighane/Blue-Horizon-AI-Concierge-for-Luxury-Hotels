```python
# run_concierge.py
from blue_horizon.services.chat_service import ConciergeOrchestrator

# Initialize the orchestrator (multi-agent manager)
orchestrator = ConciergeOrchestrator()

def run_concierge_interaction(user_input: str) -> str:
    """
    Takes user input and returns a response from the orchestrator.
    This is the core of the agent loop used in CLI and API.
    """
    try:
        response = orchestrator.handle_input(user_input)
        return response
    except Exception as e:
        return f"[Error]: {str(e)}"
```
